```python
# Importamos las librerías necesarias
from openai import OpenAI
from google.colab import userdata

# Obtenemos la llave de la pestaña "Secretos" de Colab
api_key_colab = userdata.get('OPENAI_API_KEY')
client = OpenAI(api_key=api_key_colab)

print("¡Entorno configurado correctamente!")
```

```python
def generar_recorrido_inmersia(interes_usuario, tiempo_minutos):
    """
    Utiliza Fast Prompting para generar un recorrido cultural personalizado
    en una sola llamada a la API, ahorrando costos y tiempos.
    """

    # Técnicas aplicadas: Role Prompting, Output Formatting, y Delimitadores
    prompt_sistema = """
    Eres el asistente inteligente de INMERSIA, experto en museología y cultura de Buenos Aires.
    Tu objetivo es diseñar un micro-recorrido cultural basado en el interés del usuario y su tiempo disponible.

    INSTRUCCIONES ESTRICTAS:
    1. Mantén un tono entusiasta, accesible y profesional.
    2. La respuesta debe estar formateada OBLIGATORIAMENTE en JSON.
    3. El JSON debe contener las siguientes claves:
       - "bienvenida": Un mensaje breve.
       - "paradas": Una lista con 2 o 3 puntos de interés recomendados.
       - "tip_inmersivo": Un consejo para disfrutar más la exhibición.
    """

    prompt_usuario = f"""
    Genera un recorrido para un visitante con los siguientes parámetros:
    <intereses> {interes_usuario} </intereses>
    <tiempo_disponible> {tiempo_minutos} minutos </tiempo_disponible>
    """

    try:
        respuesta = client.chat.completions.create(
            model="gpt-4o-mini",
            messages=[
                {"role": "system", "content": prompt_sistema},
                {"role": "user", "content": prompt_usuario}
            ],
            temperature=0.3, # Temperatura baja para mantener el formato estricto
            response_format={"type": "json_object"} # Forzamos la salida en JSON
        )
        return respuesta.choices[0].message.content
    except Exception as e:
        return f"Error en la consulta: {e}"

# Prueba de la POC
print("Generando recorrido para visitante de prueba...")
```
